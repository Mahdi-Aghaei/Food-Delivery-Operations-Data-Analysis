```python
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

from scipy.stats import spearmanr, pearsonr

pd.set_option("display.max_rows", 300)
pd.set_option("display.max_columns", 200)
pd.set_option("display.float_format", lambda x: f"{x:,.3f}")

RANDOM_STATE = 42


ANALYSIS_START = pd.Timestamp("2025-08-01")
ANALYSIS_END = pd.Timestamp("2025-11-01")


MIN_DISTRICT_ORDERS_FOR_RANKING = 200

def find_file(*filenames):
    for filename in filenames:
        for p in [
            Path(filename),
            Path.cwd() / filename,
            Path("/mnt/data") / filename
        ]:
            if p.exists():
                return p
    raise FileNotFoundError(f"Could not find any of: {filenames}")

details_path = find_file("order_details.csv", "order_details(3).csv")
timestamps_path = find_file("order_timestamp.csv", "order_timestamp(3).csv")

details = pd.read_csv(details_path)
timestamps = pd.read_csv(timestamps_path)

print("order_details:", details.shape)
print("order_timestamp:", timestamps.shape)
```

```python
for col in ["start_time_slot", "accept_time", "arrived_pickup_at"]:
    timestamps[col] = pd.to_datetime(
        timestamps[col],
        errors="coerce"
    )

details["created_date"] = pd.to_datetime(
    details["created_date"],
    errors="coerce"
)

df = details.merge(
    timestamps[
        [
            "order_id",
            "start_time_slot",
            "accept_time",
            "arrived_pickup_at"
        ]
    ],
    on="order_id",
    how="left",
    validate="one_to_one"
)

df = df[
    (df["start_time_slot"] >= ANALYSIS_START)
    & (df["start_time_slot"] < ANALYSIS_END)
].copy()

df = df.reset_index(drop=True)

print(f"Rows in Aug–Oct analysis period: {len(df):,}")
print(f"Unique orders: {df['order_id'].nunique():,}")
```

```python
df["BATR_raw_min"] = (
    df["arrived_pickup_at"]
    - df["start_time_slot"]
).dt.total_seconds() / 60

df["Acceptance_raw_min"] = (
    df["accept_time"]
    - df["start_time_slot"]
).dt.total_seconds() / 60

df["Arrival_raw_min"] = (
    df["arrived_pickup_at"]
    - df["accept_time"]
).dt.total_seconds() / 60



df["BATR_min"] = df["BATR_raw_min"].clip(lower=0)
df["Acceptance_min"] = df["Acceptance_raw_min"].clip(lower=0)
df["Arrival_min"] = df["Arrival_raw_min"].clip(lower=0)

df["BATR15"] = np.where(
    df["BATR_min"].notna(),
    (df["BATR_min"] <= 15).astype(int),
    np.nan
)


df["timing_complete"] = (
    df["start_time_slot"].notna()
    & df["accept_time"].notna()
    & df["arrived_pickup_at"].notna()
)

df["sequential_timing"] = (
    df["timing_complete"]
    & (df["accept_time"] >= df["start_time_slot"])
    & (df["arrived_pickup_at"] >= df["accept_time"])
)

df["arrived_before_accept"] = (
    df["timing_complete"]
    & (df["arrived_pickup_at"] < df["accept_time"])
)


seq = df[
    df["sequential_timing"]
    & df["BATR15"].notna()
].copy()

seq["BATR_component_sum"] = (
    seq["Acceptance_min"]
    + seq["Arrival_min"]
)

seq["identity_error"] = (
    seq["BATR_min"]
    - seq["BATR_component_sum"]
)

print(f"Valid BATR labels: {df['BATR15'].notna().sum():,}")
print(f"Sequential timing cohort: {len(seq):,}")
print(
    "Arrived-before-accept cohort:",
    f"{df['arrived_before_accept'].sum():,}",
    f"({df['arrived_before_accept'].mean()*100:.2f}%)"
)
print(
    "Maximum additive identity error in sequential cohort:",
    seq["identity_error"].abs().max()
)
```

```python
quality_summary = pd.DataFrame({
    "Metric": [
        "Orders in period",
        "Valid BATR",
        "Sequential timing",
        "Arrived before accept",
        "Missing start time",
        "Missing accept time",
        "Missing arrival time"
    ],
    "Count": [
        len(df),
        df["BATR15"].notna().sum(),
        df["sequential_timing"].sum(),
        df["arrived_before_accept"].sum(),
        df["start_time_slot"].isna().sum(),
        df["accept_time"].isna().sum(),
        df["arrived_pickup_at"].isna().sum()
    ]
})

quality_summary["Share of period (%)"] = (
    quality_summary["Count"] / len(df) * 100
)

display(quality_summary)
```

```python
timing_by_district = (
    df.groupby(["order_city", "district"])
      .agg(
          Orders=("order_id", "count"),
          Sequential=("sequential_timing", "sum"),
          Arrived_Before_Accept=("arrived_before_accept", "sum")
      )
      .reset_index()
)

timing_by_district["Sequential Coverage (%)"] = (
    timing_by_district["Sequential"]
    / timing_by_district["Orders"]
    * 100
)

timing_by_district["Arrived Before Accept (%)"] = (
    timing_by_district["Arrived_Before_Accept"]
    / timing_by_district["Orders"]
    * 100
)

display(
    timing_by_district.sort_values(
        "Sequential Coverage (%)"
    )
)
```

```python
def trimmed_mean(series, lower=.01, upper=.99):
    s = series.dropna()
    if len(s) == 0:
        return np.nan
    lo = s.quantile(lower)
    hi = s.quantile(upper)
    return s[(s >= lo) & (s <= hi)].mean()

def summarize_components(grouped):
    return (
        grouped.agg(
            Orders=("order_id", "count"),
            BATR15_Rate=("BATR15", "mean"),

            Acceptance_Mean=("Acceptance_min", "mean"),
            Acceptance_TrimmedMean=("Acceptance_min", trimmed_mean),
            Acceptance_Median=("Acceptance_min", "median"),
            Acceptance_P75=("Acceptance_min", lambda s: s.quantile(.75)),
            Acceptance_P90=("Acceptance_min", lambda s: s.quantile(.90)),
            Acceptance_P95=("Acceptance_min", lambda s: s.quantile(.95)),

            Arrival_Mean=("Arrival_min", "mean"),
            Arrival_TrimmedMean=("Arrival_min", trimmed_mean),
            Arrival_Median=("Arrival_min", "median"),
            Arrival_P75=("Arrival_min", lambda s: s.quantile(.75)),
            Arrival_P90=("Arrival_min", lambda s: s.quantile(.90)),
            Arrival_P95=("Arrival_min", lambda s: s.quantile(.95)),

            BATR_Mean=("BATR_min", "mean"),
            BATR_Median=("BATR_min", "median"),
            BATR_P90=("BATR_min", lambda s: s.quantile(.90)),
            BATR_P95=("BATR_min", lambda s: s.quantile(.95))
        )
        .reset_index()
    )

city_components = summarize_components(
    seq.groupby("order_city")
)

city_components["BATR15 (%)"] = (
    city_components["BATR15_Rate"] * 100
)

display(city_components)
```

```python
city_base = city_components.set_index("order_city")

benchmark = district_components.copy()

for metric in [
    "Acceptance_Mean",
    "Acceptance_TrimmedMean",
    "Acceptance_Median",
    "Acceptance_P90",
    "Acceptance_P95",
    "Arrival_Mean",
    "Arrival_TrimmedMean",
    "Arrival_Median",
    "Arrival_P90",
    "Arrival_P95",
    "BATR_Mean",
    "BATR_Median",
    "BATR_P90"
]:
    benchmark[f"City_{metric}"] = (
        benchmark["order_city"]
        .map(city_base[metric])
    )

    benchmark[f"{metric}_Gap_vs_City"] = (
        benchmark[metric]
        - benchmark[f"City_{metric}"]
    )

benchmark["Acceptance_P90_GapPct_vs_City"] = (
    benchmark["Acceptance_P90_Gap_vs_City"]
    / benchmark["City_Acceptance_P90"]
    * 100
)

benchmark["Arrival_P90_GapPct_vs_City"] = (
    benchmark["Arrival_P90_Gap_vs_City"]
    / benchmark["City_Arrival_P90"]
    * 100
)

benchmark["Primary_Tail_Deterioration"] = np.where(
    (
        benchmark["Acceptance_P90_Gap_vs_City"] <= 0
    )
    & (
        benchmark["Arrival_P90_Gap_vs_City"] <= 0
    ),
    "Neither component worse than city",
    np.where(
        benchmark["Acceptance_P90_GapPct_vs_City"]
        > benchmark["Arrival_P90_GapPct_vs_City"],
        "Acceptance tail",
        "Arrival tail"
    )
)

display(
    benchmark[
        [
            "order_city",
            "district",
            "Orders",
            "BATR15 (%)",

            "Acceptance_Median",
            "City_Acceptance_Median",
            "Acceptance_Median_Gap_vs_City",
            "Acceptance_P90",
            "City_Acceptance_P90",
            "Acceptance_P90_Gap_vs_City",
            "Acceptance_P90_GapPct_vs_City",

            "Arrival_Median",
            "City_Arrival_Median",
            "Arrival_Median_Gap_vs_City",
            "Arrival_P90",
            "City_Arrival_P90",
            "Arrival_P90_Gap_vs_City",
            "Arrival_P90_GapPct_vs_City",

            "Primary_Tail_Deterioration"
        ]
    ].sort_values(
        ["order_city", "BATR15 (%)"]
    )
)
```

```python
def process_diagnosis(row):

    a_impact = row[
        "Acceptance Normalization Impact (pp)"
    ]

    r_impact = row[
        "Arrival Normalization Impact (pp)"
    ]

    a_tail = row[
        "Acceptance_P90_GapPct_vs_City"
    ]

    r_tail = row[
        "Arrival_P90_GapPct_vs_City"
    ]

    if (
        pd.isna(a_impact)
        or pd.isna(r_impact)
    ):
        return "Insufficient component data"



    if a_impact > r_impact:
        primary = "Acceptance-stage deterioration"
    elif r_impact > a_impact:
        primary = "Post-acceptance arrival deterioration"
    else:
        primary = "Similar Acceptance / Arrival opportunity"


    if (
        pd.notna(a_tail)
        and pd.notna(r_tail)
        and a_tail > 0
        and r_tail > 0
    ):
        primary += " (both tails above city baseline)"

    return primary


rca["Evidence-Supported Process Issue"] = (
    rca.apply(
        process_diagnosis,
        axis=1
    )
)

display(
    rca[
        [
            "order_city",
            "district",
            "Orders",
            "Late_Orders",
            "BATR15 (%)",

            "Acceptance_P90_Gap_vs_City",
            "Arrival_P90_Gap_vs_City",

            "Acceptance Normalization Impact (pp)",
            "Arrival Normalization Impact (pp)",

            "Acceptance_Dev_P90",
            "Arrival_Dev_P90",

            "Evidence-Supported Process Issue"
        ]
    ].sort_values(
        ["order_city", "BATR15 (%)"]
    )
)
```

```python
city_exclusive_median = (
    district_biker_structure
    .groupby("order_city")[
        "Exclusive Biker Share (%)"
    ]
    .median()
    .to_dict()
)

def root_cause_hypothesis(row):

    issue = row[
        "Evidence-Supported Process Issue"
    ]

    city = row["order_city"]

    exclusive = row[
        "Exclusive Biker Share (%)"
    ]

    city_exclusive = city_exclusive_median.get(
        city,
        np.nan
    )

    if issue.startswith(
        "Acceptance-stage"
    ):
        hypothesis = (
            "Assignment / supply-pressure hypothesis: "
            "orders are spending unusually long before acceptance."
        )

        missing = (
            "Active bikers by district/time; offer attempts; "
            "offer recipients; rejection timestamps; biker position "
            "when offers are sent."
        )

    elif issue.startswith(
        "Post-acceptance"
    ):
        hypothesis = (
            "Post-assignment mobility / positioning hypothesis: "
            "accepted bikers are taking unusually long to reach restaurants."
        )

        if (
            pd.notna(exclusive)
            and pd.notna(city_exclusive)
            and exclusive > city_exclusive
        ):
            hypothesis += (
                " The district also has a more isolated biker pool "
                "than the city median, which strengthens the need to "
                "investigate local positioning / mobility structure."
            )

        missing = (
            "Biker GPS at acceptance; biker-to-restaurant distance; "
            "ETA; traffic; route history; concurrent assignments."
        )

    elif issue.startswith(
        "Similar"
    ):
        hypothesis = (
            "Mixed process hypothesis: both pre-acceptance and "
            "post-acceptance stages require investigation."
        )

        missing = (
            "Active biker supply + offer logs + biker GPS/ETA "
            "to restaurant."
        )

    else:
        hypothesis = (
            "No reliable process diagnosis from current timing data."
        )

        missing = (
            "Validate timestamp semantics and collect missing "
            "operational process data."
        )

    return pd.Series(
        [hypothesis, missing]
    )


rca[
    [
        "Operational Root-Cause Hypothesis",
        "Data Needed to Confirm"
    ]
] = rca.apply(
    root_cause_hypothesis,
    axis=1
)

display(
    rca[
        [
            "order_city",
            "district",
            "Orders",
            "Late_Orders",
            "BATR15 (%)",
            "Share of All Late Orders (%)",

            "Evidence-Supported Process Issue",
            "Operational Root-Cause Hypothesis",
            "Data Needed to Confirm",

            "Sequential Coverage (%)",
            "Exclusive Biker Share (%)"
        ]
    ].sort_values(
        "Late_Orders",
        ascending=False
    )
)
```

```python
context_df = valid_batr.copy()

context_df["month_ctx"] = (
    context_df["start_time_slot"]
    .dt.to_period("M")
    .astype(str)
)

context_df["weekday_ctx"] = (
    context_df["start_time_slot"]
    .dt.day_name()
)

context_df["hour_ctx"] = (
    context_df["start_time_slot"]
    .dt.hour
)

context_cols = [
    "order_city",
    "month_ctx",
    "weekday_ctx",
    "hour_ctx"
]


context_totals = (
    context_df.groupby(context_cols)
    .agg(
        Context_N=("BATR15", "size"),
        Context_Success=("BATR15", "sum")
    )
    .reset_index()
)


district_context = (
    context_df.groupby(
        context_cols + ["district"]
    )
    .agg(
        DistrictContext_N=("BATR15", "size"),
        DistrictContext_Success=("BATR15", "sum")
    )
    .reset_index()
)

district_context = district_context.merge(
    context_totals,
    on=context_cols,
    how="left"
)

district_context["LOO_N"] = (
    district_context["Context_N"]
    - district_context["DistrictContext_N"]
)

district_context["LOO_Success"] = (
    district_context["Context_Success"]
    - district_context["DistrictContext_Success"]
)


district_context["LOO_Valid"] = (
    district_context["LOO_N"] >= 30
)

district_context["LOO_Rate"] = np.where(
    district_context["LOO_Valid"],
    district_context["LOO_Success"]
    / district_context["LOO_N"],
    np.nan
)

district_context["Expected_Success"] = (
    district_context["DistrictContext_N"]
    * district_context["LOO_Rate"]
)

district_context["Expected_Variance"] = (
    district_context["DistrictContext_N"]
    * district_context["LOO_Rate"]
    * (1 - district_context["LOO_Rate"])
)

valid_context_rows = district_context[
    district_context["LOO_Valid"]
].copy()

context_benchmark = (
    valid_context_rows.groupby(
        ["order_city", "district"]
    )
    .agg(
        Covered_Orders=("DistrictContext_N", "sum"),
        Actual_Success=("DistrictContext_Success", "sum"),
        Expected_Success=("Expected_Success", "sum"),
        Expected_Variance=("Expected_Variance", "sum")
    )
    .reset_index()
)

total_district_orders = (
    context_df.groupby(["order_city", "district"])
    .size()
    .rename("Total_Orders")
    .reset_index()
)

context_benchmark = context_benchmark.merge(
    total_district_orders,
    on=["order_city", "district"],
    how="left"
)

context_benchmark["Context Coverage (%)"] = (
    context_benchmark["Covered_Orders"]
    / context_benchmark["Total_Orders"]
    * 100
)

context_benchmark["Observed Context-Matched BATR15 (%)"] = (
    context_benchmark["Actual_Success"]
    / context_benchmark["Covered_Orders"]
    * 100
)

context_benchmark["Expected BATR15 from Rest of City (%)"] = (
    context_benchmark["Expected_Success"]
    / context_benchmark["Covered_Orders"]
    * 100
)

context_benchmark["Context Gap (pp)"] = (
    context_benchmark["Observed Context-Matched BATR15 (%)"]
    - context_benchmark["Expected BATR15 from Rest of City (%)"]
)


context_benchmark["Context Standardized Gap"] = (
    context_benchmark["Actual_Success"]
    - context_benchmark["Expected_Success"]
) / np.sqrt(
    context_benchmark["Expected_Variance"]
)

def context_status(gap):
    if pd.isna(gap):
        return "Insufficient comparable context"
    if gap < -2:
        return "Local underperformance"
    if gap > 2:
        return "Outperforms comparable city contexts"
    return "Near city-context baseline"

context_benchmark["Context Interpretation"] = (
    context_benchmark["Context Gap (pp)"]
    .apply(context_status)
)

display(
    context_benchmark.sort_values(
        "Context Gap (pp)"
    )
)
```

```python
def refined_component_pattern(row):

    median_gap = (
        row["Acceptance_Median_LateMinusSuccess"]
        - row["Arrival_Median_LateMinusSuccess"]
    )

    p90_gap = (
        row["Acceptance_P90_LateMinusSuccess"]
        - row["Arrival_P90_LateMinusSuccess"]
    )

    total_a = row[
        "Acceptance Share of Total Late Minutes (%)"
    ]

    if median_gap > 2:
        typical = "Acceptance-typical"
    elif median_gap < -2:
        typical = "Arrival-typical"
    else:
        typical = "Mixed-typical"

    if p90_gap > 2:
        tail = "Acceptance-tail"
    elif p90_gap < -2:
        tail = "Arrival-tail"
    else:
        tail = "Mixed-tail"

    if pd.notna(total_a):
        if total_a > 55:
            total = "Acceptance dominates total minutes"
        elif total_a < 45:
            total = "Arrival dominates total minutes"
        else:
            total = "Total minutes roughly balanced"
    else:
        total = "No total-minute estimate"

    return f"{typical} / {tail} / {total}"

component_refined["Refined Component Pattern"] = (
    component_refined.apply(
        refined_component_pattern,
        axis=1
    )
)

display(
    component_refined[
        [
            "order_city",
            "district",
            "Orders_BATR15_0",

            "Acceptance_Median_LateMinusSuccess",
            "Arrival_Median_LateMinusSuccess",

            "Acceptance_P90_LateMinusSuccess",
            "Arrival_P90_LateMinusSuccess",

            "Acceptance Share of Total Late Minutes (%)",
            "Arrival Share of Total Late Minutes (%)",

            "Median Acceptance Share per Late Order (%)",
            "Median Arrival Share per Late Order (%)",

            "Refined Component Pattern"
        ]
    ].sort_values(
        "Orders_BATR15_0",
        ascending=False
    )
)
```

```python
refined_rca = location_priority.merge(
    component_refined[
        [
            "order_city",
            "district",

            "Acceptance_Median_LateMinusSuccess",
            "Arrival_Median_LateMinusSuccess",

            "Acceptance_P90_LateMinusSuccess",
            "Arrival_P90_LateMinusSuccess",

            "Acceptance Share of Total Late Minutes (%)",
            "Arrival Share of Total Late Minutes (%)",

            "Median Acceptance Share per Late Order (%)",
            "Median Arrival Share per Late Order (%)",

            "Refined Component Pattern"
        ]
    ],
    on=["order_city", "district"],
    how="left"
)

refined_rca = refined_rca.merge(
    district_adjusted[
        [
            "order_city",
            "district",
            "Acceptance_Dev_Median",
            "Acceptance_Dev_P90",
            "Arrival_Dev_Median",
            "Arrival_Dev_P90"
        ]
    ],
    on=["order_city", "district"],
    how="left"
)

refined_rca = refined_rca.merge(
    load_comparison[
        [
            "order_city",
            "district",
            "High minus Low BATR15 (pp)",
            "Acceptance Dev Change High-Low (min)",
            "Arrival Dev Change High-Low (min)"
        ]
    ],
    on=["order_city", "district"],
    how="left"
)

refined_rca = refined_rca.merge(
    vendor_breadth,
    on=["order_city", "district"],
    how="left"
)

refined_rca = refined_rca.merge(
    district_biker_structure[
        [
            "order_city",
            "district",
            "Exclusive Biker Share (%)",
            "Orders_per_Unique_Biker"
        ]
    ],
    on=["order_city", "district"],
    how="left"
)

refined_rca = refined_rca.merge(
    timing_by_district[
        [
            "order_city",
            "district",
            "Sequential Coverage (%)",
            "Arrived Before Accept (%)"
        ]
    ],
    on=["order_city", "district"],
    how="left"
)

def final_rca_interpretation(row):

    context_gap = row["Context Gap (pp)"]

    a_med = row[
        "Acceptance_Median_LateMinusSuccess"
    ]
    r_med = row[
        "Arrival_Median_LateMinusSuccess"
    ]

    a_p90 = row[
        "Acceptance_P90_LateMinusSuccess"
    ]
    r_p90 = row[
        "Arrival_P90_LateMinusSuccess"
    ]


    if pd.notna(context_gap) and context_gap > 2:
        local = "Not a local underperformance; district outperforms comparable city contexts"
    elif pd.notna(context_gap) and context_gap >= -2:
        local = "Near comparable city-context baseline"
    else:
        local = "Local underperformance"


    med_diff = a_med - r_med
    tail_diff = a_p90 - r_p90

    if med_diff > 2 and tail_diff >= -2:
        stage = "Acceptance-dominant"
    elif med_diff < -2 and tail_diff <= 2:
        stage = "Arrival-dominant"
    elif med_diff > 2 and tail_diff < -2:
        stage = "Acceptance-typical with Arrival tail risk"
    elif med_diff < -2 and tail_diff > 2:
        stage = "Arrival-typical with Acceptance tail risk"
    else:
        stage = "Mixed Acceptance / Arrival"

    return f"{local} | {stage}"

refined_rca["Refined RCA Diagnosis"] = (
    refined_rca.apply(
        final_rca_interpretation,
        axis=1
    )
)

display(
    refined_rca.sort_values(
        "Late_Orders",
        ascending=False
    )
)
```

```python
def operational_hypothesis_v2(row):

    context_gap = row["Context Gap (pp)"]
    diagnosis = row["Refined RCA Diagnosis"]

    load_batr_gap = row.get(
        "High minus Low BATR15 (pp)",
        np.nan
    )

    load_a_change = row.get(
        "Acceptance Dev Change High-Low (min)",
        np.nan
    )

    load_r_change = row.get(
        "Arrival Dev Change High-Low (min)",
        np.nan
    )

    vendor_a = row.get(
        "Pct_Vendors_Acceptance_P90_Above_City",
        np.nan
    )

    vendor_r = row.get(
        "Pct_Vendors_Arrival_P90_Above_City",
        np.nan
    )

    exclusive = row.get(
        "Exclusive Biker Share (%)",
        np.nan
    )

    if pd.notna(context_gap) and context_gap > 2:
        return pd.Series([
            "High late volume is mainly a scale/context issue rather than local underperformance.",
            "No immediate local intervention from this table alone; monitor city-wide drivers and volume."
        ])

    if pd.notna(context_gap) and context_gap >= -2:
        context_note = (
            "District is near the city-context baseline; "
            "treat the component pattern as structural/descriptive rather than proof of local underperformance. "
        )
    else:
        context_note = ""

    if "Acceptance-dominant" in diagnosis:

        if (
            pd.notna(load_batr_gap)
            and load_batr_gap <= -5
            and pd.notna(load_a_change)
            and pd.notna(load_r_change)
            and load_a_change > load_r_change
        ):
            hypothesis = (
                context_note
                + "Acceptance deterioration becomes materially worse in high-load 15-minute intervals "
                "even after City-Hour adjustment. This supports an assignment/capacity-pressure hypothesis."
            )
        else:
            hypothesis = (
                context_note
                + "Acceptance is the dominant abnormal stage, but observed order volume alone does not fully explain it. "
                "Investigate offer flow, active supply, rejection/ignore behavior, pricing/incentives, and biker positioning before acceptance."
            )

        data_needed = (
            "Active bikers by district/time; offer attempts and recipients; reject/ignore events; "
            "biker GPS at offer; offered compensation/incentive."
        )

    elif "Arrival-dominant" in diagnosis:

        broad = (
            pd.notna(vendor_r)
            and vendor_r >= 70
        )

        isolated = (
            pd.notna(exclusive)
            and exclusive >= 40
        )

        if broad and isolated:
            hypothesis = (
                context_note
                + "Arrival deterioration is broad across vendors and the biker pool is unusually isolated. "
                "This supports a localized post-assignment mobility/positioning hypothesis rather than a single-restaurant problem."
            )
        elif broad:
            hypothesis = (
                context_note
                + "Arrival deterioration is broad across vendors, suggesting an area-level post-assignment movement issue."
            )
        else:
            hypothesis = (
                context_note
                + "Arrival is the dominant abnormal stage. Investigate biker-to-restaurant travel and whether the issue is concentrated in specific vendors."
            )

        data_needed = (
            "Biker GPS at acceptance; biker-to-restaurant distance; ETA; traffic; actual route; "
            "previous delivery endpoint; concurrent assignments."
        )

    else:
        hypothesis = (
            context_note
            + "Both process stages contribute materially. A single-stage explanation is not supported."
        )

        data_needed = (
            "Active supply + offer funnel + biker GPS/ETA to restaurant."
        )

    return pd.Series([
        hypothesis,
        data_needed
    ])

refined_rca[
    [
        "Refined Operational Hypothesis",
        "Additional Data Needed"
    ]
] = refined_rca.apply(
    operational_hypothesis_v2,
    axis=1
)

display(
    refined_rca[
        [
            "order_city",
            "district",
            "Orders",
            "Late_Orders",
            "BATR15 (%)",
            "Share of All Late Orders (%)",

            "Context Gap (pp)",
            "Context Interpretation",

            "Acceptance_Median_LateMinusSuccess",
            "Arrival_Median_LateMinusSuccess",

            "Acceptance_P90_LateMinusSuccess",
            "Arrival_P90_LateMinusSuccess",

            "Acceptance Share of Total Late Minutes (%)",
            "Arrival Share of Total Late Minutes (%)",

            "High minus Low BATR15 (pp)",
            "Acceptance Dev Change High-Low (min)",
            "Arrival Dev Change High-Low (min)",

            "Pct_Vendors_Acceptance_P90_Above_City",
            "Pct_Vendors_Arrival_P90_Above_City",

            "Exclusive Biker Share (%)",
            "Sequential Coverage (%)",

            "Refined RCA Diagnosis",
            "Refined Operational Hypothesis",
            "Additional Data Needed"
        ]
    ].sort_values(
        ["Context Gap (pp)", "Late_Orders"],
        ascending=[True, False]
    )
)
```

```python
city_change_explanation = monthly_change[
    monthly_change["Delta_BATR_Mean"].notna()
].copy()

def explain_month_change(row):

    delta_b = row["Delta_BATR_Mean"]
    delta_a = row["Delta_Acceptance_Mean"]
    delta_r = row["Delta_Arrival_Mean"]

    if delta_b <= 0:
        return (
            "Mean BATR improved or stayed flat; "
            f"Acceptance change={delta_a:+.2f}m, "
            f"Arrival change={delta_r:+.2f}m."
        )

    if abs(delta_a) + abs(delta_r) == 0:
        return "No material component movement."

    if delta_a > 0 and delta_r > 0:
        a_share = delta_a / delta_b * 100
        r_share = delta_r / delta_b * 100
        return (
            f"BATR worsened by {delta_b:+.2f}m: "
            f"Acceptance contributed {delta_a:+.2f}m ({a_share:.1f}%), "
            f"Arrival contributed {delta_r:+.2f}m ({r_share:.1f}%)."
        )

    if delta_r > 0 and delta_a <= 0:
        return (
            f"BATR worsened by {delta_b:+.2f}m despite Acceptance improving "
            f"({delta_a:+.2f}m); deterioration came from Arrival ({delta_r:+.2f}m)."
        )

    if delta_a > 0 and delta_r <= 0:
        return (
            f"BATR worsened by {delta_b:+.2f}m despite Arrival improving "
            f"({delta_r:+.2f}m); deterioration came from Acceptance ({delta_a:+.2f}m)."
        )

    return "Mixed change."

city_change_explanation["RCA Explanation"] = (
    city_change_explanation.apply(
        explain_month_change,
        axis=1
    )
)

display(
    city_change_explanation[
        [
            "order_city",
            "month",
            "BATR15 (%)",
            "Delta_BATR_Mean",
            "Delta_Acceptance_Mean",
            "Delta_Arrival_Mean",
            "RCA Explanation"
        ]
    ]
)
```

```python
final_rca_table = refined_rca[
    [
        "order_city",
        "district",
        "Orders",
        "Late_Orders",
        "BATR15 (%)",
        "Share of All Late Orders (%)",

        "Expected BATR15 from Rest of City (%)",
        "Context Gap (pp)",
        "Context Interpretation",

        "Acceptance_Median_LateMinusSuccess",
        "Arrival_Median_LateMinusSuccess",

        "Acceptance_P90_LateMinusSuccess",
        "Arrival_P90_LateMinusSuccess",

        "Acceptance Share of Total Late Minutes (%)",
        "Arrival Share of Total Late Minutes (%)",

        "High minus Low BATR15 (pp)",
        "Acceptance Dev Change High-Low (min)",
        "Arrival Dev Change High-Low (min)",

        "Pct_Vendors_Acceptance_P90_Above_City",
        "Pct_Vendors_Arrival_P90_Above_City",

        "Exclusive Biker Share (%)",
        "Sequential Coverage (%)",

        "Refined Component Pattern",
        "Refined RCA Diagnosis",
        "Refined Operational Hypothesis",
        "Additional Data Needed"
    ]
].copy()



final_rca_table["Local Underperformance Magnitude (pp)"] = (
    (-final_rca_table["Context Gap (pp)"])
    .clip(lower=0)
)

final_rca_table["Local Impact Score"] = (
    final_rca_table["Late_Orders"]
    * final_rca_table[
        "Local Underperformance Magnitude (pp)"
    ]
)

final_rca_table = final_rca_table.sort_values(
    [
        "Local Impact Score",
        "Late_Orders"
    ],
    ascending=False
)

display(final_rca_table)
```
